```python
# question - 3
def reverse(num):
  rev = 0
  while (num>0):
    digit = num%10
    rev = rev*10 + digit
    num = num // 10
  return rev
reverse(1234)
```

```python
# question - 4
def count_digits(num):
  count = 0
  while (num>0):
    num = num // 10
    count +=1
  return count
count_digits(45678927312369)
```

```python
# question 5
def ispalindrome(num):
  temp = num
  rev = 0
  while (num>0):
    digit = num%10
    rev = rev*10 + digit
    num = num // 10
  if (rev == temp):
    print("palindrome")
  else :
    print("not palindrome")

ispalindrome(4567)
ispalindrome(424)
```

```python
# question - 7
n1=int(input("enter the number"))
n2=int(input("enter the number"))
def sum(a,b):
  return a+b
def substrat (a,b):
  return a-b
def multiply(a,b):
  return a*b
def divide(a,b):
  return a/b
print("1. ADDITION")
print("2. SUBSTRACTION")
print("3. MULTIPLICATION")
print("4. DIVISION")
choice=int(input("enter the choice  "))
if choice==1:
  print(sum(n1,n2))
elif choice==2:
  print(substrat(n1,n2))
elif choice==3:
  print(multiply(n1,n2))
elif choice==4:
  print(divide(n1,n2))
else:
  print("invalid choice")
```
